1.	Crie uma lista preenchida com os 5 primeiros colocados de um Campeonato de Futebol, na ordem de colocação. Depois mostre:

a. Apenas os 3 primeiros colocados;

b. Os últimos 2 colocados;

c. Uma lista com os times em ordem alfabética;

d. Em que posição da tabela se encontra o Barcelona;

```python
#fazendo a lista com os 5 primeiros (colocando o Barcelona no meio pro item D dar certo)
times = ["Real Madrid", "Barcelona", "Atletico Madrid", "Girona", "Sevilla"]

print("Tabela original:", times)
print("-" * 30)

# a) pegando os 3 primeiros -> slicing
print("a) Top 3:", times[:3])

# b) pegando os 2 últimos
print("b) Últimos 2 colocados:", times[-2:])

# c) ordem alfabética (usando sorted pra não bagunçar a lista original)
times_alfabetica = sorted(times)
print("c) Ordem alfabética:", times_alfabetica)

# d) achando a posição do Barcelona
#a função index() acha onde ele tá (lembrando que começa no zero, então somo 1 pra mostrar a posição certa)
posicao_barca = times.index("Barcelona") + 1
print(f"d) O Barcelona está na {posicao_barca}ª posição da tabela")
```

2.	Crie dois conjuntos, um para cada loja. Identifique quais modelos de smartphones cada uma delas vendem. Em seguida, mostre quais modelos no total você terá opção de comprar se visita-las e quais modelos se encontram disponíveis em ambas as lojas;

```python
# criando os conjuntos (sets) para cada loja
loja1 = {"iPhone 13", "Galaxy S23", "Moto G54", "Redmi Note 12"}
loja2 = {"Galaxy S23", "iPhone 15", "Poco X5", "Moto G54"}

# mostrando o que cada loja vende separadamente
print("Modelos vendidos na Loja 1:", loja1)
print("Modelos vendidos na Loja 2:", loja2)
print("-" * 40)

# juntando tudo para ver as opções totais
# obs: o '|' junta os dois sets e já tira as repetições automaticamente
total_opcoes = loja1 | loja2

print("Total de opções visitando as duas lojas:")
print(total_opcoes)
print("-" * 40)

# olhando o que tem em comum nas duas lojas
# obs: o '&' pega apenas quem tá presente na loja1 E na loja2
em_ambas = loja1 & loja2

print("Modelos disponíveis em ambas as lojas:")
print(em_ambas)
```

3.	Faça um programa que leia o nome e a média de um aluno e guarde-os em um dicionário. Em seguida, a partir da média (para ser aprovado deve ter média >=50), gere a situação final do aluno ('AP' ou 'RP'), que também deve ser guardada neste dicionário. No final, mostre todo o conteúdo deste dicionário;

```python
# cria o dicionário vazio pra guardar os dados
aluno = {}

# lendo os dados
aluno['nome'] = input("Nome do aluno: ")
aluno['media'] = float(input("Média do aluno: "))

# ifzinho pra verificar se passou (AP) ou reprovou (RP)
if aluno['media'] >= 50:
    aluno['situacao'] = 'AP'
else:
    aluno['situacao'] = 'RP'

print("-" * 30)

print("Conteúdo do dicionário:")
print(aluno)

print("-" * 30)
```

4. Faça um programa que leia o nome e peso de 3 pessoas e no final mostre o nome da pessoa mais pesada e a mais leve;

```python
nome_pesada = ""
maior_peso = 0
nome_leve = ""
menor_peso = 0

# loop pra ler os dados das 3 pessoas
for i in range(3):
    nome = input("Digite o nome: ")
    peso = float(input("Digite o peso: "))

    # na primeira volta, a primeira pessoa é a mais pesada e a mais leve
    if i == 0:
        nome_pesada = nome
        maior_peso = peso
        nome_leve = nome
        menor_peso = peso
    else:
        # compara com os pesos guardados pra ver se atualiza
        if peso > maior_peso:
            maior_peso = peso
            nome_pesada = nome

        if peso < menor_peso:
            menor_peso = peso
            nome_leve = nome

print("Pessoa mais pesada:", nome_pesada)
print("Pessoa mais leve:", nome_leve)
```

5. Desenvolva um programa que leia o nome, idade e sexo de n pessoas. No final, mostre: a. A média de idade do grupo; b. Quantas mulheres têm menos de 20 anos.

 Dica: em Python, os operadores booleanos básicos são and, or e not.

```python
n = int(input("Quantas pessoas você quer ler? "))

soma_idades = 0
mulheres_sub_20 = 0

for i in range(n):
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    sexo = input("Sexo [M/F]: ").upper()

    soma_idades = soma_idades + idade

    if sexo == 'F' and idade < 20:
        mulheres_sub_20 = mulheres_sub_20 + 1

media_idade = soma_idades / n

print("a. Média de idade do grupo:", media_idade)
print("b. Mulheres com menos de 20 anos:", mulheres_sub_20)
```

7. Usando a lista de ingredientes já preenchida para a receita de bolo, agora crie mais dois conjuntos representando os ingredientes que duas pessoas diferentes têm em casa. Mostre quais ingredientes da receita ainda faltam comprar pelas pessoas para se fazer o bolo.

```python
# lista baseada no resultado do exercício anterior
ingredientes = ['farinha', 'açúcar', 'manteiga', 'leite', 'fermento']

# converte pra set pra facilitar a conta
receita = set(ingredientes)

# cria os conjuntos do que cada um tem
pessoa1 = {'farinha', 'açúcar'}
pessoa2 = {'leite', 'ovo'}

#junta o que os dois têm
tem_juntos = pessoa1 | pessoa2

# diminui pra ver o que faltou pra receita
falta_comprar = receita - tem_juntos

print(falta_comprar)
```

8. Crie um dicionário para armazenar os dados de um produto (nome, preço e quantidade em estoque). Peça ao usuário os dados de 3 produtos diferentes e guarde cada dicionário em uma lista. No final, percorra a lista e mostre, para cada produto, seu nome e o valor total em estoque (preço × quantidade).

```python
produtos = []

for i in range(3):
    produto = {}
    produto['nome'] = input("Nome do produto: ")
    produto['preco'] = float(input("Preço: "))
    produto['quantidade'] = int(input("Quantidade: "))
    produtos.append(produto)

for p in produtos:
    valor_total = p['preco'] * p['quantidade']
    print(p['nome'], "-", valor_total)
```
