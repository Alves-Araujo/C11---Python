1. Apresente a porcentagem de missões que deram certo

```python
import pandas as pd

# lendo o arquivo (coloquei o sep=';' porque esse csv é separado por ponto e vírgula)
df = pd.read_csv('space.csv', sep=';')

# pegando o total de linhas
total = len(df)

# filtrando só as que deram certo e pegando a quantidade
sucessos = len(df[df['Status Mission'] == 'Success'])

# calculando a porcentagem
porcentagem = (sucessos / total) * 100

print(f"Porcentagem de missões que deram certo: {porcentagem:.2f}%")
```

2. Qual a media de gastos de uma missão especial se baseando em missões que possuam valores disponíveis (> 0)?

```python
import pandas as pd

df = pd.read_csv('space.csv', sep=';')

# o nome da coluna no csv tem um espaço na frente (' Cost')
# filtrando só as missões que têm o valor maior que zero
missoes_com_valor = df[df[' Cost'] > 0]

# calculando a média dessa galera
media_gastos = missoes_com_valor[' Cost'].mean()

print(f"A média de gastos é: {media_gastos:.2f}")
```

3. Encontre quantas missões espaciais neste Dataset foram realizadas pelos Estados Unidos (EUA)

```python
import pandas as pd

df = pd.read_csv('space.csv', sep=';')

# filtrando a coluna Location pra ver se contém 'USA' no texto
missoes_eua = df[df['Location'].str.contains('USA')]

# contando quantas linhas sobraram no filtro
total_eua = len(missoes_eua)

print("Total de missões realizadas pelos EUA:", total_eua)
```

4. Encontre qual foi a missão mais cara realizada pela empresas “SpaceX”

```python
import pandas as pd

df = pd.read_csv('space.csv', sep=';')

# filtra o dataframe pra deixar só a SpaceX
spacex = df[df['Company Name'] == 'SpaceX']

# descobre qual foi o valor máximo gasto por eles (lembrando do espaço no nome da coluna)
maior_custo = spacex[' Cost'].max()

# pega a linha exata que tem esse custo máximo
missao_mais_cara = spacex[spacex[' Cost'] == maior_custo]

print("Missão mais cara da SpaceX:")
# mostrando só o detalhe da missão e o valor pra não poluir a tela com a tabela toda
print(missao_mais_cara[['Detail', ' Cost']])
```

5. Mostre o nome das empresas que já realizaram missões espaciais, juntamente com suas respectivas quantidades de missões (use o for no final para mostrar as informações)

```python
import pandas as pd

df = pd.read_csv('space.csv', sep=';')

# o value_counts já agrupa e conta quantas vezes cada empresa aparece na coluna
contagem_empresas = df['Company Name'].value_counts()

# usando o for no final como pede o enunciado
for empresa, quantidade in contagem_empresas.items():
    print(empresa, "-", quantidade, "missões")
```

6. Qual a porcentagem de missões realizadas com foguetes cujo status é "StatusRetired" (coluna Status Rocket)?

```python
import pandas as pd

df = pd.read_csv('space.csv', sep=';')

# pegando o total de missões da tabela
total_missoes = len(df)

# filtrando pra ver quantos têm o status aposentado
aposentados = len(df[df['Status Rocket'] == 'StatusRetired'])

# calculando a porcentagem
porcentagem = (aposentados / total_missoes) * 100

print(f"Porcentagem de foguetes com status 'StatusRetired': {porcentagem:.2f}%")
```

7. Quantas missões foram lançadas a partir de localizações que contêm "Russia" (coluna Location)?

```python
import pandas as pd

df = pd.read_csv('space.csv', sep=';')

# filtrando a coluna Location pra achar a palavra Russia
missoes_russia = df[df['Location'].str.contains('Russia')]

# contando quantas linhas sobraram no filtro
total_russia = len(missoes_russia)

print("Total de missões lançadas da Rússia:", total_russia)
```
