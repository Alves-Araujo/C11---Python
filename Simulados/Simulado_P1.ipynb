Questão 2:

Conte e em seguida mostre quais são as diferentes Regiões do planeta segundoeste dataset;

Questão 3:

Mostre qual a taxa média de alfabetização (Literacy (%)) do planeta segundo este dataset;

```python
#
alfabetizacao = dataset[1: , 9:10].astype(float) #atribui os valores selecionados na variável alfabetizacao e transforma o valor para um flutuante
soma = np.sum(alfabetizacao)
media =  soma / len(alfabetizacao)
print(media)
```

Questão 4:

Conte quantos países são da América do Norte (NORTHERN AMERICA)
segundo este dataset;

```python
regiao = dataset[1: , 1:2]
print(np.sum(np.char.find(regiao, 'NORTHERN AMERICA') >= 0))
```

Questao 5:

Encontre qual país da América do Sul e Caribe (LATIN AMER. & CARIB)
possui a maior renda per capita (GDP ($ per capita));

```python
paise = dataset[1:][np.char.find(dataset[1:,1],'LATIN AMER. & CARIB') >= 0]
GDP = paise[:,8:9].astype(float)

print(paise[np.argmax(GDP), 0])
```
